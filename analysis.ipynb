```python
# Step 1 — Basic overview of our data
print("Total posts:", len(df))
print("Date range:", df['Date'].min().date(), "to", df['Date'].max().date())
print("\nContent types:")
print(df['Content_Type'].value_counts())
print("\nTopics:")
print(df['Topic'].value_counts())
```

```python
# Average views by topic
avg_views = df.groupby('Topic')['Views'].mean().sort_values(ascending=False)

print("Average Views by Topic:")
print(avg_views.astype(int))
```

```python
import plotly.express as px

fig = px.bar(
    x=avg_views.index,
    y=avg_views.values,
    title="Average Views by Content Topic — Shri Balaji Entertainment",
    labels={'x': 'Topic', 'y': 'Average Views'},
    color=avg_views.values,
    color_continuous_scale='Oranges'
)

fig.show()
```

```python
# Extract day of week from date
df['Day_of_Week'] = df['Date'].dt.day_name()

# Average views by day
avg_views_day = df.groupby('Day_of_Week')['Views'].mean().sort_values(ascending=False)

print("Average Views by Day of Week:")
print(avg_views_day.astype(int))
```

```python
# Order days correctly
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
             'Friday', 'Saturday', 'Sunday']

fig2 = px.bar(
    x=avg_views_day.reindex(day_order).index,
    y=avg_views_day.reindex(day_order).values,
    title="Best Days to Post — Average Views by Day",
    labels={'x': 'Day', 'y': 'Average Views'},
    color=avg_views_day.reindex(day_order).values,
    color_continuous_scale='Blues'
)

fig2.show()
```

```python
# Calculate engagement rate
df['Engagement_Rate'] = ((df['Likes'] + df['Comment']) / df['Views'] * 100).round(2)

# Average engagement rate by topic
avg_engagement = df.groupby('Topic')['Engagement_Rate'].mean().round(2).sort_values(ascending=False)

print("Average Engagement Rate by Topic (%):")
print(avg_engagement)
```

```python
fig3 = px.bar(
    x=avg_engagement.index,
    y=avg_engagement.values,
    title="Engagement Rate by Topic (%) — Shri Balaji Entertainment",
    labels={'x': 'Topic', 'y': 'Engagement Rate (%)'},
    color=avg_engagement.values,
    color_continuous_scale='Greens'
)

fig3.show()
```

```python
# Monthly views trend
df['Month'] = df['Date'].dt.to_period('M').astype(str)

monthly_views = df.groupby('Month')['Views'].sum().reset_index()

fig4 = px.line(
    monthly_views,
    x='Month',
    y='Views',
    title="Monthly Views Trend — Shri Balaji Entertainment",
    labels={'Month': 'Month', 'Views': 'Total Views'},
    markers=True
)

fig4.show()
```

```python
# Summary table
summary = df.groupby('Topic').agg(
    Total_Posts=('Topic', 'count'),
    Avg_Views=('Views', 'mean'),
    Avg_Likes=('Likes', 'mean'),
    Avg_Engagement_Rate=('Engagement_Rate', 'mean')
).round(2).sort_values('Avg_Views', ascending=False)

print("=== Shri Balaji Entertainment — Content Performance Summary ===")
print(summary.to_string())
```
