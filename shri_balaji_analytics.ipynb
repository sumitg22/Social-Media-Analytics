```python
# Step 1 — Basic overview of our data
print("Total posts:", len(df))
print("Date range:", df['Date'].min().date(), "to", df['Date'].max().date())
print("\nContent types:")
print(df['Content_Type'].value_counts())
print("\nTopics:")
print(df['Topic'].value_counts())
```

```python
# Average views by topic
avg_views = df.groupby('Topic')['Views'].mean().sort_values(ascending=False)

print("Average Views by Topic:")
print(avg_views.astype(int))
```

```python
import plotly.express as px

fig = px.bar(
    x=avg_views.index,
    y=avg_views.values,
    title="Average Views by Content Topic — Shri Balaji Entertainment",
    labels={'x': 'Topic', 'y': 'Average Views'},
    color=avg_views.values,
    color_continuous_scale='Oranges'
)

fig.show()
```
